Separate features and target

```python
X = df.drop(
    "Heart Disease",
    axis=1
)

y = df["Heart Disease"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X = pd.get_dummies(
    X,
    drop_first=True
)

print(X.head())

print(
    "Number of features after encoding:",
    X.shape[1]
)
```

Convert Boolean columns to integers

```python
X = X.astype(int)

print(X.head())
```

Train-test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

train logistic regression

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(max_iter=1000, random_state=42)
logistic_model.fit(X_train, y_train)

y_pred_lr = logistic_model.predict(X_test)
print("Logistic Regression model trained successfully!")
```

Train Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight='balanced'
)

random_forest_model.fit(X_train, y_train)

y_pred_rf = random_forest_model.predict(X_test)
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)

y_pred_knn = knn_model.predict(X_test)
print("KNN model trained successfully!")
```

Model Evaluation

```python
def evaluate_model(model_name, y_test, y_pred):

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    print("\n==============================")
    print(model_name)
    print("==============================")

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

    print("\nClassification Report:")
    print(classification_report(
        y_test,
        y_pred,
        zero_division=0
    ))

    return accuracy, precision, recall, f1
```

Run Evaluation

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

lr_results = evaluate_model(
    "Logistic Regression",
    y_test,
    y_pred_lr
)

rf_results = evaluate_model(
    "Random Forest",
    y_test,
    y_pred_rf
)

knn_results = evaluate_model(
    "KNN",
    y_test,
    y_pred_knn
)
```

Compare Models

```python

results = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Random Forest',
        'KNN'
    ],

    'Accuracy': [
        lr_results[0],
        rf_results[0],
        knn_results[0]
    ],

    'Precision': [
        lr_results[1],
        rf_results[1],
        knn_results[1]
    ],

    'Recall': [
        lr_results[2],
        rf_results[2],
        knn_results[2]
    ],

    'F1 Score': [
        lr_results[3],
        rf_results[3],
        knn_results[3]
    ]
})

print(results)
```

Comparison graph

```python
results.set_index('Model').plot(
    kind='bar',
    figsize=(10,6)
)

plt.title("Machine Learning Model Comparison")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(loc='lower right')
plt.show()
```

Select the best Model


```python
model_dict = {
    "Logistic Regression": logistic_model,
    "Random Forest": random_forest_model,
    "KNN": knn_model
}

best_model_name = results.loc[
    results["F1 Score"].idxmax(),
    "Model"
]

best_model = model_dict[best_model_name]

print("Best Model:", best_model_name)

print(
    "Best Accuracy:",
    round(
        results.loc[
            results["Model"] == best_model_name,
            "Accuracy"
        ].iloc[0] * 100,
        2
    ),
    "%"
)
```

```python
age = int(input("Enter Age: "))
gender = input("Enter Gender (Male/Female): ")

while True:
    cholesterol_input = input("Enter Cholesterol: ").strip()
    try:
        cholesterol = int(cholesterol_input)
        break
    except ValueError:
        print("Please enter a whole-number cholesterol value.")

blood_pressure = int(
    input("Enter Blood Pressure: ")
)

heart_rate = int(
    input("Enter Heart Rate: ")
)

smoking = input(
    "Enter Smoking (Never/Current/Former): "
)

alcohol = input(
    "Enter Alcohol Intake (Heavy/Moderate): "
)

exercise_hours = int(
    input("Enter Exercise Hours: ")
)

family_history = input(
    "Enter Family History (Yes/No): "
)

diabetes = input(
    "Enter Diabetes (Yes/No): "
)

obesity = input(
    "Enter Obesity (Yes/No): "
)

stress_level = int(
    input("Enter Stress Level: ")
)

while True:
    blood_sugar_input = input("Enter Blood Sugar (number): ").strip()
    try:
        blood_sugar = int(blood_sugar_input)
        break
    except ValueError:
        print("Please enter a whole-number blood sugar value.")

angina = input(
    "Enter Exercise Induced Angina (Yes/No): "
)

chest_pain = input(
    "Enter Chest Pain Type: "
)
```
